```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
import matplotlib.pyplot as plt
```

```python
# Step 1: Create a circuit with 1 qubit and 1 classical bit
qc = QuantumCircuit(1, 1)

# Step 2: Put the qubit in superposition (like a swirled ice cream mix)
qc.h(0)

# Step 3: Measure the qubit (forcing it to choose chocolate or vanilla)
qc.measure(0, 0)
```

```python
# Step 4: Simulate the experiment
simulator = Aer.get_backend("aer_simulator")
job = simulator.run(qc, shots=100)
result = job.result()
counts = result.get_counts(qc)
print(counts)
```

```python
# Step 5: Display the results
labels = ["Chocolate (0)", "Vanilla (1)"]
values = [counts.get("0", 0), counts.get("1", 0)]

plt.bar(labels, values, color=["saddlebrown", "gold"])
plt.xlabel("Flavor")
plt.ylabel("Count")
plt.title("Quantum Ice Cream Machine Results")
plt.show()
```

```python
from qiskit.visualization import plot_histogram

job = simulator.run(qc, shots=1000)
counts = job.result().get_counts(qc)
print(counts)
plot_histogram(counts)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector

qc_state = QuantumCircuit(1)
qc_state.h(0)

state = Statevector.from_instruction(qc_state)
print(state)
print(state.probabilities())
plot_bloch_multivector(state)
```
