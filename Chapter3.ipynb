```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
```

```python
# Load CIFAR-10 dataset (10 categories of images)
(X_train, y_train), (X_test, y_test) = keras.datasets.cifar10.load_data()
```

```python
# Normalize pixel values
X_train, X_test = X_train / 255.0, X_test / 255.0
```

```python
model = keras.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(32,32,3)),
    layers.MaxPooling2D(2,2),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')  # 10 classes
])
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=1, validation_data=(X_test, y_test))
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test)
print("Test Accuracy:", test_acc)
```

```python
import torch
import torch.nn as nn
import torchvision.transforms
import torchvision.datasets as datasets
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import torchvision.transforms as transforms
```

```python
transform = transforms.Compose([transforms.ToTensor(),
                                transforms.Normalize((0.5,), (0.5,))])
```

```python
train_data = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
test_data = datasets.MNIST(root='./data', train=False, download=True, transform=transform)
```

```python
train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader = DataLoader(test_data, batch_size=64, shuffle=False)
```

```python
class CNN(nn.Module):
    def __init__(self):
        super(CNN, self).__init__()
        self.conv1 = nn.Conv2d(1, 32, kernel_size=3, stride=1, padding=1)
        self.pool = nn.MaxPool2d(2, 2)
        self.fc1 = nn.Linear(32 * 14 * 14, 10)

    def forward(self, x):
        x = self.pool(torch.relu(self.conv1(x)))
        x = x.view(-1, 32 * 14 * 14)
        x = self.fc1(x)
        return x
```

```python
model = CNN()

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
for epoch in range(5):
    for images, labels in train_loader:
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
```

```python
correct = 0
total = 0
with torch.no_grad():
    for images, labels in test_loader:
        outputs = model(images)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print(f"Test Accuracy: {100 * correct / total:.2f}%")
```

```python
alpha = 0.1        # Learning rate
gamma = 0.99       # Discount factor
epsilon = 1.0      # Exploration rate
epsilon_decay = 0.995
epsilon_min = 0.01
episodes = 2000    # Number of games
```

```python
import gym
import gymnasium as gym
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import random
import numpy as np
from collections import deque
```

```python
import warnings
warnings.simplefilter(action='ignore', category=DeprecationWarning)
```

```python
env = gym.make("CartPole-v1")
state_size = env.observation_space.shape[0]
action_size = env.action_space.n
```

```python
class DQN(nn.Module):
    def __init__(self, state_size, action_size):
        super(DQN, self).__init__()
        self.fc1 = nn.Linear(state_size, 24)
        self.fc2 = nn.Linear(24, 24)
        self.fc3 = nn.Linear(24, action_size)

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        return self.fc3(x)
```

```python
model = DQN(state_size, action_size)
optimizer = optim.Adam(model.parameters(), lr=0.001)
loss_function = nn.MSELoss()
```

```python
memory = deque(maxlen=2000)

gamma = 0.95         # Discount factor
epsilon = 1.0        # Exploration rate
epsilon_min = 0.01
epsilon_decay = 0.995
batch_size = 64
episodes = 1000
```

```python
def remember(state, action, reward, next_state, done):
    memory.append((state, action, reward, next_state, done))
```

```python
def replay():
    if len(memory) < batch_size:
        return

    minibatch = random.sample(memory, batch_size)
    for state, action, reward, next_state, done in minibatch:
        target = reward
        if not done:
            target += gamma * torch.max(model(torch.FloatTensor(next_state))).item()

        output = model(torch.FloatTensor(state))[action]

        # Ensure target has the same shape as output
        loss = loss_function(output.unsqueeze(0), torch.tensor([target], dtype=torch.float32))

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
```

```python
for episode in range(episodes):
    state = env.reset()[0]
    done = False
    total_reward = 0

    while not done:
        if np.random.rand()< epsilon:
            action = random.randrange(action_size)  # Explore
        else:
            action = torch.argmax(model(torch.FloatTensor(state))).item()  # Exploit

        next_state, reward, done, _, _ = env.step(action)  # For gym&gt;=0.26; for older versions, use: next_state, reward, done, _ = env.step(action)
        remember(state, action, reward, next_state, done)
        state = next_state
        total_reward += reward

        replay()
        epsilon = max(epsilon * epsilon_decay, epsilon_min)

    print(f"Episode {episode+1}, Total Reward: {total_reward}")

print("Training complete!")
```
